## Topic 1: Mutual Information & Chi-Square Test (Feature Selection)

You have a dataset with many features (columns). Not all features are useful for predicting your target. Feature selection helps you figure out which features actually matter.

**Mututal Information**

Simple analogy: Imagine you're trying to guess if it will rain tomorrow. If I tell you "it's cloudy today", that helps your guess. If I tell you "your shirt is blue today", that doesn't help at all.
Mutual Information measures exactly this — how much does knowing feature X help you predict target Y?

- Score = 0 -> feature is completely useless 
- Higer Score -> feature is more useful

It works for both classification and regression, and it handles non-linear relationships too, which correlation matrix can't do.

```python
from sklearn.feature_selection import mutual_info_classif
scores = mutual_info_classif(X, y)
# gives each feature a score — higher = more important
```

**Chi Square Test**

Simple analogy: You want to know if gender affects survival on the Titanic. Chi-Square asks — "is the pattern I'm seeing real, or just random chance?"

- It works only with categorical features and a categorical target
- It gives a p-value — if p < 0.05, the feature is statistically significant

```python
from sklearn.feature_selection import chi2
scores, p_values = chi2(X_categorical, y)
# low p-value = feature is important
```

## Topic 2: Principal Component Analysis (PCA)

You have a dataset with many features. PCA helps you reduce the number of features while keeping as much information as possible.

**Simple Analogy:** Imagine you have 10 different photos of the same person from slightly different angles. Instead of storing all 10, you pick the 2-3 photos that best capture what the person looks like. You lose a tiny bit of detail but save a lot of space — and the photos still look like the same person. PCA does exactly this with features.

**What does PCA actually do:**

Say you have 10 features. PCA:
- Looks at all 10 features together 
- Finds new "combined" features called Principal Components (PCs)
- PC1 captures the most variance in data 
- PC2 captures the second most, and so on 
- You keep only the top 2-3 PCs instead of all 10

Variance = how spread out / informative the data is

**Example Code:**
```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Reduce to 2 components
pca = PCA(n_components=2)
X_reduced = pca.fit_transform(X_scaled)  # always scale before PCA!

# Check how much variance is explained
print(pca.explained_variance_ratio_)
# e.g. [0.60, 0.25] → PC1=60%, PC2=25%

# Scree plot
plt.plot(pca.explained_variance_ratio_)
plt.xlabel('Principal Component')
plt.ylabel('Variance Explained')
plt.show()

# 2D scatter plot
plt.scatter(X_reduced[:, 0], X_reduced[:, 1], c=y)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.show()
```

## Topic 3: t-SNE — t-Distributed Stochastic Neighbor Embedding

Like PCA, t-SNE also reduces high-dimensional data to 2D for visualization. But it works very differently and is purely for visualization — not for feeding into ML models.

**PCA vs t-SNE - The Core Difference**

PCA is like taking a photograph of a 3D object — it finds the best angle to preserve the overall structure and distances.
t-SNE is like a map of neighborhoods — it cares more about keeping similar points close together and dissimilar points far apart. It's great at revealing clusters.

**Simple Analogy**

Imagine 1000 students with grades in 10 subjects. You want to plot them in 2D.
- PCA will spread them out to show maximum variance — good for seeing the overall spread
- t-SNE will group them into clusters — "science kids here, arts kids there, average kids in the middle"

**How It Works**
- In high dimensions, it measures how similar each point is to every point (closer = more similar)
- It then tries to recreate those same relationships in 2D
- It pulls similar points together and pushes dissimilar points apart
- It does this iteratively, like physically moving points around until clusters form naturally

**Example Code:**
```python
from sklearn.manifold import TSNE

tsne = TSNE(n_components=2, perplexity=30, n_iter=1000, random_state=42)
X_tsne = tsne.fit_transform(X_scaled)

# Visualize
import matplotlib.pyplot as plt
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='viridis')
plt.title('t-SNE Visualization')
plt.xlabel('t-SNE 1')
plt.ylabel('t-SNE 2')
plt.colorbar()
plt.show()
```
